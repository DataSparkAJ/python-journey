```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,15,5,20]

plt.plot(x,y)
plt.show()
```

```python
x = ["Mon", "Tue", "Wed", "Thu"]
y = [15,20,18,22]
plt.plot(x,y)
plt.xlabel("Day of the Week")
plt.ylabel("Sale per day")
plt.title("Bakery sales this week!")
plt.show()
```

```python

# plt.plot(x,y, color = 'color name', linestyle = 'line_style', linewidth = value, marker = 'marker symbol', label = 'label_name')
month = [1,2,3,4]
sales = [1000,1500,1200,1800]
plt.plot(month,sales, color = 'red', linestyle = '--', linewidth = 2, marker = 'o', label = 'sales report')
plt.xlabel('Months')
plt.ylabel('sales per month')
plt.legend(loc = 'upper left', fontsize = 10)
plt.grid(linestyle = ':', color = 'blue',linewidth = 1)
plt.title('Monthly sales data report')
plt.xlim(1,4)
plt.ylim(0,2000)
plt.xticks([1,2,3,4],['M1','M2','M3','M4'])
plt.show()

```

```python
# plt.bar(x,height,color='color name',label='label name')
product = ['A','B','C','D']
sales = [1000,1500,800,1200]
plt.bar(product,sales, color = 'orange', label = 'sales 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title("Product Sale Comparision")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
# plt.pie(values, labels = label_list, color = color_list, autopct = '%1.1f%%')

regions = ['North', 'South', 'East', 'West']
revenue = [1000,1500,1300,1900]

plt.pie(revenue, labels = regions, colors = ['gold', 'lightblue', 'pink', 'orange'], autopct= '%1.1f%%')
plt.title('Revenue Contribution By Region')
plt.legend()
plt.show()
```

```python
#  plt.hist(data, bins = number_of_bins, color = 'color_name', edgecolor = 'black')
import matplotlib.pyplot as plt

scores = [45,56,75,46,58,78,65,79,98,76,56,78,97,76,88,80,57]

plt.hist(scores, bins = 5, color = 'yellow', edgecolor = 'black')
plt.title('Score Distribution of Students')
plt.xlabel('Score Range')
plt.ylabel('No of Students')
plt.show()
```

```python
# plt.scatter(x,y, color = 'color_name', marker = 'marker style', label = 'label name')
import matplotlib.pyplot as plt

hours_studied = [1,2,3,4,5,6,7,8]
marks_scored = [50,55,60,65,70,75,80,85]

# scatter plot
plt.scatter(hours_studied, marks_scored, color = 'red', marker = 'o', label = 'Students Data')
plt.title('Relationship between hours studied and marks scored')
plt.ylabel('Exam Score')
plt.xlabel('Hours Studied')
plt.legend()
plt.grid()
plt.show()
```

```python
# plt.scatter(x,y, color = 'color_name', marker = 'marker style', label = 'label name')
import matplotlib.pyplot as plt

plt.scatter([1,2,3], [50,60,70], color = 'blue', label = 'Class A')
plt.scatter([1,2,3], [55,58,79], color = 'red', label = 'Class B')
plt.title('Comparision of Two Classes')
plt.ylabel('Exam Score')
plt.xlabel('Hours Studied')
plt.legend()
plt.grid()
plt.show()
```

```python
# plt.subplot(nrow, ncol, index)
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,30,40]

plt.subplot(1,2,1)
plt.plot(x,y)
plt.title('Line Chart')

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Bar Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,30,40]
# fig, ax = plt.subplots(nrows, ncols, figsize =(width, height))
fig, ax = plt.subplots(1, 2, figsize = (10,5))

x = [1,2,3,4]
y = [10,20,30,40]

ax[0].plot(x,y, color = 'blue')
ax[0].set_title('Line Plot')

ax[1].bar(x,y, color = 'green')
ax[1].set_title('Bar Plot')

plt.suptitle("Comparision of Bar and Line Chart")
plt.tight_layout()
plt.show()
```
